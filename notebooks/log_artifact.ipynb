```python
import os
import torch
import torch.nn as nn
from torchvision import models
import mlflow
import mlflow.pytorch
from dotenv import load_dotenv
```

```python
load_dotenv()

model = models.resnet18(weights=None)
model.fc = nn.Linear(512, 101)

model_path = os.getenv("MODEL_PATH")

state_dict = torch.load(model_path, map_location='cpu')

model.load_state_dict(state_dict)
model.eval()
```

```python
mlflow.set_tracking_uri("https://attempt-evict-comment.ngrok-free.dev")
mlflow.set_experiment('ResNet18_Food101_Run')
with mlflow.start_run(run_id="67a7dd194d2a46a89336513f059944a0"):
    mlflow.pytorch.log_model(pytorch_model=model, artifact_path="model")
```
